```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Activation, Flatten, Conv2D, MaxPooling2D
import pickle

X = pickle.load(open("X.pickle","rb"))
y = pickle.load(open("y.pickle","rb"))

#normalize data by scaling
X = X/255.0

#first layer
model = Sequential()    #This initializes a linear stack of layers. You can imagine building a sandwich layer by layer; the data flows from the top layer straight down to the bottom.
model.add(      Conv2D(64, (3,3), input_shape = X.shape[1:])    ) #Add conv layer. 64=number of filters, 3,3 kernal = the window, input shape
model.add(Activation("relu"))
model.add(MaxPooling2D(pool_size=(2,2)))

#Second layer
model.add(Conv2D(64, (3,3))) 
model.add(Activation("relu"))
model.add(MaxPooling2D(pool_size=(2,2)))

#Flatten the layer since dense is 1D, conv is 2D
model.add(Flatten())
#final dense layer
model.add(Dense(64))    #Every neuron in this layer is connected to every neuron in the previous layer. This layer takes the high-level features learned by the CNN and attempts to combine them to make a decision.

model.add(Dense(1)) #It is only 1 neuron because it is yes/no
model.add(Activation('sigmoid'))    #utilize the sigmoid activation function

model.compile(loss="binary_crossentropy",           #The "scorecard." It calculates how wrong the model is. "Binary Crossentropy" is the standard math for checking errors in Yes/No tasks.
              optimizer = "adam",                    #The "teacher." When the model makes a mistake (loss is high), the optimizer tweaks the weights (the filter values) to improve the result
              metrics=['accuracy'])

model.fit(X, y, batch_size=32, validation_split=0.1)      #how many we procress at a time, 10% of validation out of sample test         starts the actual training

```
